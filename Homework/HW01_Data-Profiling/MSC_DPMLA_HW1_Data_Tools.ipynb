# **4. Great Expectations**
**Great Expectations (GX) is a popular open-source Python framework dedicated to data quality, data validation, and automated profiling.** It allows data engineers and data scientists to express exactly what they expect from their data using a declarative syntax, protecting pipelines from bad data.

```python
pip install great_expectations
```

```python
import great_expectations as gx
import pandas as pd
```

```python
context = gx.get_context()
```

```python
# Create Data Source & Batch from df
data_source = context.data_sources.add_pandas("pandas_source")
data_asset = data_source.add_dataframe_asset(name="claude_baseline")
batch_definition = data_asset.add_batch_definition_whole_dataframe("full_df")
batch = batch_definition.get_batch(batch_parameters={"dataframe": df})
```

```python
#Create Expectation Suite
suite = context.suites.add(gx.ExpectationSuite(name="claude_baseline_suite"))
```

```python
# Add Expectations (Rules)
# "id" should not be null (Completeness)
suite.add_expectation(gx.expectations.ExpectColumnValuesToNotBeNull(column="id"))
suite.add_expectation(gx.expectations.ExpectColumnValuesToBeUnique(column="id"))

# Accuracy of "age"
suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(column="age", min_value=18, max_value=100)
)

# Validity of "sex"
suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeInSet(column="sex", value_set=[0, 1])
)

# "dm_duration" should be possitive number
suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(column="dm_duration", min_value=0, max_value=None)
)

# Validity of "bmi"
suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(column="bmi", min_value=10, max_value=60)
)

# "baseline_hba1c" should be apporpiate.
suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(column="baseline_hba1c", min_value=3, max_value=20)
)

# "baseline_egfr" should be apporpiate.
suite.add_expectation(
    gx.expectations.ExpectColumnValuesToBeBetween(column="baseline_egfr", min_value=0, max_value=150)
)

# Importance col. shouldn't have missing value
for col in ["age", "sex", "dm_duration", "bmi", "baseline_hba1c", "baseline_egfr"]:
    suite.add_expectation(gx.expectations.ExpectColumnValuesToNotBeNull(column=col))
```

```python
#Run validation result
validation_result = batch.validate(suite)
print(validation_result)
```

```python
for r in validation_result.results:
    print(r["expectation_config"]["type"], "-", r["success"])
```
